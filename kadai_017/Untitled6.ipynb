```python
class Human:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print("大人です。")
        else:
            print("大人ではありません。")



person1 = Human("田中", 25)
person2 = Human("佐藤", 18)


people = [person1, person2]


for person in people:
    person.check_adult()


```
