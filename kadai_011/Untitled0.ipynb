```python
import random

sum = 0

while sum < 20:

  num = random.randint(1, 10)

  print(f"{num}が出ました。")

  if num % 2 == 0:

      print(f"偶数なので加算しません。")

      continue

  sum = sum + num

  print(f"現在の合計は{sum}です。")

```
