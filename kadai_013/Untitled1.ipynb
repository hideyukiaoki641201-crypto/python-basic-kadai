```python
def greet_user(user_name: str) -> str:
    if user_name == "":
        return "こんにちは、ゲストさん！"
    else:
        return f"こんにちは、{user_name}さん！"


print(greet_user("青木"))
print(greet_user("田中"))
print(greet_user(""))
```

```python
def calculate_tax(price, tax_rate):
    total = price + (price * tax_rate / 100)
    return total

result = calculate_tax(110, 10)

print(f"税込金額は{result}円です。")
```
